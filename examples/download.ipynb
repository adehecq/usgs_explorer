# Connection to the API

Credentials for the Earth Explorer portal can be obtained [here](https://ers.cr.usgs.gov/register/). Note that you need to specify specifically all datasets you plan to access through the API.


```python
username = os.getenv("USGS_USERNAME") or input("Enter your USGS username: ")
token = os.getenv("USGS_TOKEN") or input("Enter your USGS token: ")

api = usgsxplore.API(username, token)
```

```python
dataset = "declassii"
entity_ids = ["DZB1216-500280L001001", "DZB1216-500280L006001"] 
label = "123456"

# product number = 0 for medium resolution
urls = list(api.get_download_links(dataset, entity_ids, label=label))
```

# Download aerial high resolution images

Here we will look at an example of somes [aerial images](https://www.usgs.gov/centers/eros/science/usgs-eros-archive-aerial-photography-aerial-photo-single-frames?qt-science_center_objects=0#qt-science_center_objects) take the 1978-09-06 on arizona.

```python
dataset = "aerial_combin"
bbox = [-111.87783, 32.66157, -111.60280, 32.89378]
date_interval = ("1978-09-06", "1978-09-06")

aerial_scenes_metadata = api.search(dataset, bbox=bbox, date_interval=date_interval)
```

```python
# get a list of entity ids
entity_ids = [metadata["entityId"] for  metadata in aerial_scenes_metadata]

# Download only the first 4 scenes
# product_number = 0 for Medium Resolution Product
api.download(dataset, entity_ids[:4], output_dir="./1978_09_06_aerial", product_number=0, verbose=True)
```

# Download KH9 MC images

Here we will look at an example of [declassified images](https://www.usgs.gov/centers/eros/science/usgs-eros-archive-declassified-data-declassified-satellite-imagery-2?qt-science_center_objects=0#qt-science_center_objects) take the 1980-08-22 in island

```python
dataset = "declassii"
bbox = [-25.7520, 63.0960, -12.7441, 67.3070]
date_interval = ("1980-08-22", "1980-08-22")
meta_filter = 'camera=L & DOWNLOAD_AVAILABLE=Y'

kh9mc_scenes_metadata = api.search(dataset, bbox=bbox, date_interval=date_interval, meta_filter=meta_filter)
```

```python
# get a list of entity ids
entity_ids = [metadata["entityId"] for metadata in kh9mc_scenes_metadata]

# Download only the first 2 scenes (~ 3 Go)
# it take's around 3 minutes on my machine
api.download(dataset, entity_ids[:2], output_dir="./1980_08_22_kh9mc", verbose=True)
```
